```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import json

# Target article URL
article_url = "https://www.bbc.com/news/technology-66979350"

# Optional: run headless
options = Options()
# options.add_argument("--headless")  # Uncomment to run without opening the browser

# Initialize browser
driver = webdriver.Chrome(options=options)

# Open the article
driver.get(article_url)
time.sleep(5)  # Wait for content to load

# Extract headline
headline = driver.find_element(By.TAG_NAME, "h1").text.strip()

# Extract article body paragraphs
paragraphs = driver.find_elements(By.CSS_SELECTOR, 'article p')

# Combine paragraphs
article_text = "\n".join(p.text for p in paragraphs if p.text.strip())

# Save data
article_data = {
    "Headline": headline,
    "URL": article_url,
    "Content": article_text
}

# Close browser
driver.quit()

# Save to JSON
with open("bbc_article.json", "w", encoding="utf-8") as f:
    json.dump(article_data, f, ensure_ascii=False, indent=4)

print("Article data saved to 'bbc_article.json'")
```
