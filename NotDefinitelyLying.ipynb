```python
def is_definitely_lying(n, accusations):
    # Build a list of who each person is blaming 
    graph = [[] for _ in range(n)]
    for i in range(n):
        blamed = accusations[i] - 1  
        graph[blamed].append(i)      

    for i in range(n):
        blamed = accusations[i] - 1
        if accusations[blamed] - 1 == i:
            return "Definitely lying"

    return "Not Definitely lying"

# Example inputs
inputs = [
    (3, [1, 3, 1]),
    (2, [2, 1]),
    (2, [2, 2]),
    (3, [2, 3, 1]),
    (3, [1, 2, 3]),
]

# Run and print results
for n, accusation_list in inputs:
    print(is_definitely_lying(n, accusation_list))
```
